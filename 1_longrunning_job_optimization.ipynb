Q. You have a large PySpark job that reads 100M+ rows from a delta table, applies complex transformation, and writes back to another delta table. The job is running slower each day. How would you debug and optimize this in PySpark?

-------------

1. Check Spark UI:
    * Go the Spark UI in databricks job
    * Look at stages with longest execution times
    * Identify wide transformations (e.g., join, groupBy, distinct, orderBy)

------------------

2. Skew Detection
    * Check if a stage is running with 1 task takig significantly longer

--------------------------

3. Join Optimization
    * Use broadcast joins for small tables

----------------

4. Partitioning
    * check if input delta table has optimal partitioning 
    * Use df.rdd.getNumPartitions() to see what's going on.
    * Repartition wisely before wide operations.

-------------------

5. Caching and Persistence
    * Cache intermediate DataFrames if resued

---------------------------

6. File size in input Delta table:
    * If your Delta table has too many small files, use 

    OPTIMIZE delta_table_name or 
    
    OPTIMIZE delta.'table_path'

---------------------

7. Z-Ordering for Query Performance

OPTIMIZE delta_table_name ZORDER By ('col_1','col_2')

During the optimization of files, ZORDER keeps rows with same values in ZORDER columns in same or adjacent files.

ZORDER is expensive but later query will be faster
