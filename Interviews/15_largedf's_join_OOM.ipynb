You need to join two large DataFrames in PySpark. One fits in memory, but the other does not. How do you optimize the join operation to avoid a shuffle?

----

Here note - One fits in memory

Use broadcast join to avoid a shuffle. Broadcasting sends the smaller DF to all executors, allowing each executor to perform a local join with its partition of the large DataFrame. This prevents costly shuffles across the network.

-------

Spark's Catalyst Optimizer can automatically broadcast small DataFrames if they're under the

spark.sql.autoBroadcastJoinThreshold (default is 10MB), but we can enforce it manually using

large_df.join(broadcast(small_df), on='key')

-----------------

Lets assume small df of 100MB and cluster (13.xlarge) with 4 workers - each worker has 30GB

Now 100MB small df is copied to all 4 workers (not split) and then sent to executors

NOTE :- Each worker has 1 executor

What is an Executor in Spark?
A Spark executor is a JVM (Java Virtual Machine) process that runs on a worker node and is responsible for executing your code and tasks.
