While working with dataframes in pyspark, we can't directly use python raw values directly as column values

Instead they should be column values --- lit(value)

You only need lit() when you're passing a raw constant alone as a column.
When you're using it in a column expression (col(...) + 5), PySpark takes care of lit() for you.

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark = SparkSession.builder.appName('new_1').getOrCreate()

df=spark.read.csv('test_data/simple-zipcodes.csv',header=True,inferSchema=True)
df.show(5)
```

```python
''' Now adding new column country_name as United States.'''

df.withColumn('country_name',lit('United States')).show(5)

# df.withColumn('country_name','United States') - This wont work
```

```python
# rand generates a number between 0.0 to 1.0  (e.g :- 0.98, 0.43)

df_1=df.withColumn('temp_col', rand()*10)
df_1.show(5)
```

```python
# Now lets use floor 
df_1.withColumn('temp_col', floor('temp_col')).show(5)
```

NOTE :- This is rand(), floor(), lit() are helpful in salting technique to solve data skewness
