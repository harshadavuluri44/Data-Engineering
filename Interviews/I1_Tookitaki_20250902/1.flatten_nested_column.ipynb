explode() function from pyspark.sql.functions flattens nested arrays or maps (only) inside column of dataframe.

---

NOTE :- Any function we import cannot be applied directly on dataframe

Only methods are applied on df.

Use methods + functions but not df.functions

-------

```python
''' SYNTAX '''

from pyspark.sql.functions import explode, explode_outer, col

df_exploded = df.withColumn('exploded_col', explode('col'))

''' OR '''

df_exploded = df.select('id', 'name', explode('col').alias('exploded_col'))

''' explode() will ignore if exploded_col has null values or null arrays
    In this case use explode_outer'''


''' Explode a map '''

df_exploded = df.withColumn('exploded_col',explode('col')) \
                .withColumn('exploded_col_key', col('exploded_col').getField('key')) \
                .withColumn('exploded_col_value', col('exploded_col').getField('value'))

df_exploded = df.select('id', explode_outer('col').alias('key','value'))
```
